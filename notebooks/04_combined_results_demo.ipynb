```python
# 04_combined_results_demo.ipynb

import json
import pandas as pd
from IPython.display import display

# 1️⃣ Load Exa.ai sources
with open("exa_search_results.json", "r", encoding="utf-8") as f:
    exa_sources = json.load(f)

# 2️⃣ Load Gemini responses
with open("responses_log.json", "r", encoding="utf-8") as f:
    gemini_responses = json.load(f)

# 3️⃣ Optional: import memory from Notebook 03 if needed
# from notebook_03_memory_test import get_recent_memory
# recent_memory = get_recent_memory()
# For now, we'll skip it or show placeholder

# 4️⃣ Combine into a Pandas DataFrame
rows = []
for i, src in enumerate(exa_sources):
    # Pick the corresponding Gemini answer or placeholder
    ai_answer = gemini_responses.get(f"q{i+1}", "Answer placeholder")
    
    rows.append({
        "Question": src.get("title", ""),
        "Source Title": src.get("title", ""),
        "URL": src.get("url", ""),
        "Snippet": src.get("snippet", "").replace("\n", " ").strip(),
        "AI Answer": ai_answer,
        # Memory context can be added if imported from Notebook 03
        "Memory Context": "Optional memory info"
    })

df_combined = pd.DataFrame(rows)

# 5️⃣ Display the combined table
print("🔹 Combined Results Table:")
display(df_combined)
```
